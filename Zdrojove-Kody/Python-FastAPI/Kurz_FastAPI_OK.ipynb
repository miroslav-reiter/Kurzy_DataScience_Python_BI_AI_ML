# 🔌 Top knižnice pre prácu s API v Pythone
1. **🚀 FastAPI:** Moderný framework pre budovanie RESTful API. Je extrémne rýchly, podporuje asynchrónne volania a automaticky generuje dokumentáciu cez Swagger UI.
   
2. **🌐 Requests:** Jednoduchá a veľmi populárna knižnica na odosielanie HTTP požiadaviek (GET, POST, PUT, DELETE). Skvelá voľba pre prácu s externými API.
   
3. **🕸️ httpx:** Asynchrónna alternatíva ku knižnici requests. Podporuje HTTP/1.1 aj HTTP/2, timeouty, cookies a ďalšie pokročilé funkcie.
   
4. **🔐 Authlib:** Knižnica zameraná na OAuth1, OAuth2 a OpenID Connect. Vhodná na zabezpečenie autentifikácie v API aplikáciách.
   
5. **🧱 Flask:** Mikroframework na tvorbu jednoduchých API. Má veľkú komunitu a množstvo rozšírení. Môže byť základom pre malé až stredne veľké API projekty.
   
6.  **🏗️ Django REST Framework (DRF):** Silná nadstavba nad Django na tvorbu REST API. Obsahuje zabudované nástroje na autentifikáciu, serializáciu, routing a pagination.
   
7.  📦 marshmallow: Knižnica na validáciu, deserializáciu a serializáciu dát. Často sa používa v kombinácii s Flask a inými REST frameworkmi.
    
8.  **⚡ Starlette:** Ľahký asynchrónny ASGI framework, ktorý tvorí základ pre FastAPI. Je ideálny na vysokovýkonné API a mikroservisy.
    
9.  **🧾 Pydantic:** Knižnica na validáciu a konverziu dát pomocou typov v Pythone. Používa sa vo FastAPI, SQLModel a ďalších moderných frameworkoch.
    
10. **📚 OpenAPI / Swagger Tools:** Sada nástrojov na automatickú dokumentáciu API podľa štandardu OpenAPI. FastAPI a DRF ich automaticky generujú.

## 🔌 Čo je API (Application Programming Interface)?
- Je rozhranie, ktoré umožňuje komunikáciu medzi dvoma softvérovými systémami – napríklad medzi klientom a serverom.
  
- API definuje, akým spôsobom možno pristupovať k funkcionalite softvérového systému, bez nutnosti poznať jeho vnútorné detaily.
  
- V kontexte webových aplikácií sa najčastejšie stretávame s REST API, ktoré komunikuje pomocou protokolu HTTP a pracuje s dátami vo formáte JSON alebo XML.
  
- API môže byť buď interné (privátne) pre komunikáciu medzi internými komponentmi, alebo verejné (public) pre externých vývojárov, ktorí môžu nad systémom vytvárať vlastné aplikácie.

## ⚡ Čo je FastAPI?
- FastAPI je moderný webový framework pre tvorbu RESTful API v jazyku Python (**Minimálna** verzia: **Python 3.8**, **Odporúčaná** verzia: **Python 3.10** alebo novšia)
- Je postavený na Starlette (asynchrónny webový framework) a Pydantic (validácia dát)
- Podporuje asynchrónne programovanie pomocou async / await, čím je mimoriadne rýchly
- Automaticky generuje dokumentáciu API pomocou OpenAPI (Swagger) a ReDoc
- Výborne sa integruje s typovaním v Pythone vďaka podpore type hints

# 🧩 Hlavné komponenty FastAPI
**🧮 Pydantic** (na validáciu modelov)
- Zabezpečuje validáciu a serializáciu dát pomocou typových anotácií.
- Používa sa na definovanie vstupných dát (BaseModel) – automaticky overuje typy a generuje chyby.
- Základ pre formuláre, JSON vstupy, a odpovede v API.

**⚡ Starlette** (na HTTP routovanie a odpovede)
- Asynchrónny webový framework určený pre rýchle API.
- FastAPI je na ňom postavené a využíva jeho smerovanie, middleware a request/response objekty.
- Starlette umožňuje jednoduché použitie WebSocket, Background Tasks, Mountovanie atď.

**🚀 Uvicorn**
- ASGI server, ktorý spúšťa aplikáciu FastAPI.
- Je extrémne rýchly, podporuje async/await, HTTP/2 a WebSockety.
- Slúži ako vstupný bod – spúšťa sa cez uvicorn main:app --reload.

**📘 OpenAPI**
- Špecifikácia, ktorá definuje štandard pre dokumentáciu REST API.
- FastAPI automaticky generuje dokumentáciu (Swagger UI a ReDoc) na základe typov a modelov.
- Vývojári môžu testovať API interaktívne bez písania dokumentácie manuálne.

## 2️⃣a Vytvorenie základného modelu v Pydantic
**📌 Úloha:** Vytvoriť jednoduchý dátový model Uzivatel pomocou knižnice Pydantic, ktorý overí typy vstupných údajov.

- Model bude odvodený od triedy BaseModel
- Bude obsahovať dve polia: meno ako reťazec (str) a vek ako celé číslo (int)
- Pri vytvorení inštancie sa automaticky overí typová správnosť vstupu
- Slúži ako základ pre validáciu údajov v aplikáciách FastAPI
- Tento príklad pomáha pochopiť, ako Pydantic zabezpečuje bezpečné a predvídateľné spracovanie dát.

**🧪 Otestujte si:**
správny vstup: Pouzivatel(meno="Miro", vek=30)
chybný vstup: Pouzivatel(meno="Miro", vek="tridsať") → vyhodí chybu ValidationError

**🔗 Dokumentácia:**
- https://docs.pydantic.dev/latest/
- https://docs.pydantic.dev/latest/concepts/models/
- https://fastapi.tiangolo.com/tutorial/body/

```python
from pydantic import BaseModel

class Pouzivatel(BaseModel):
    meno: str
    vek: int

dataOK = Pouzivatel(meno="Miro", vek=30)
# dataNG = Pouzivatel(meno="Miro", vek="tridsať")
print(dataOK)

try:
    Pouzivatel(meno=123, vek="abc")
except Exception as e:
    print(e)
```

```python
# Výchozie hodnoty + nepovinné pole
# 📦 cena má výchoziu hodnotu, popis je nepovinný.
from typing import Optional

# Produkt je model (trieda) odvodený z BaseModel
# Takto môže FastAPI/Pydantic automaticky:
# validovať vstupy
#spracovať JSON z API
# vrátiť objekt s typmi a metódami (model_dump(), .nazov, atď.)
class Produkt(BaseModel):
    nazov: str
    cena: float = 9.99
    popis: Optional[str] = None

print(Produkt(nazov="Notebook"))

# Vstup ako dict → výstup ako objekt
# ✅ Veľmi vhodné pre FastAPI – príjem dát vo forme JSON/dict.
vstup = {"nazov": "Myš", "cena": 19.5}
# Rozbalenie slovníka do argumentov modelu
# Operátor ** rozbalí slovník ako kľúčové argumenty funkcie/konstruktora
# Je to skratka za: Produkt(nazov="Myš", cena=19.5)
produkt = Produkt(**vstup)
print(produkt.nazov)

# Výstup ako dict (napr. späť do JSON)
# Od verzie Pydantic 2.0 sa pre zjednotenie s .model_validate() zaviedlo .model_dump()
print(produkt.model_dump())

# toto je už zastaralé .dict() síce ešte funguje, ale odstraňuje sa vo verzii 3.0
# Funkcia .dict() bola v Pydantic 1.x zaužívaná na export modelu ako slovníka
#print(produkt.dict())
```

```python
# PyCharm FastAPI Projekt
# Importovanie triedy FastAPI z balíka fastapi
from fastapi import FastAPI

# Vytvorenie inštancie aplikácie FastAPI
app = FastAPI()

# Definovanie GET HTTP metódy na koreňovej URL "/"
@app.get("/")
async def root():
    # Návratová hodnota vo formáte JSON
    return {"message": "Hello World"}

# Definovanie GET HTTP metódy na URL "/hello/{name}"
# {name} je dynamický parameter, ktorý bude súčasťou URL
@app.get("/hello/{name}")
async def say_hello(name: str):
    # Návratová hodnota s personalizovanou správou
    return {"message": f"Hello {name}"}
```

**📦 Sekcia Schemas**
FastAPI využíva Pydantic na validáciu a automaticky vytvára tieto schémy:

**📄 HTTPValidationError**
Popis chybového objektu v prípade nevalidného vstupu cez HTTP.

Obsahuje zoznam chýb v rôznych poliach.

**📄 ValidationError**
Vnútorná štruktúra jednej validačnej chyby (napr. zlé meno, nevyplnený parameter atď.).

**📑 Využitie**
- Rozhranie Swagger UI Vám umožňuje:
- Testovať endpointy interaktívne
- Vidieť typy požiadaviek a odpovedí
- Skúšať rôzne hodnoty bez použitia externého klienta (napr. Postman)

```python
! pip install fastapi uvicorn
# Prvý endpoint - GET a POST metóda
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Item(BaseModel):
    name: str
    price: float

@app.get("/")
def read_root():  # definícia funkcie endpointu
    return {"message": "Vitajte vo FastAPI"}  # návratová hodnota

@app.post("/items/")
def create_item(item: Item):  # definícia funkcie endpointu
    return {"item_name": item.name, "item_price": item.price}  # návratová hodnota
```

## 3️⃣ Spustenie FastAPI servera v Jupyter Notebooku a testovanie GET/POST požiadaviek
📌 Úloha: Spustiť FastAPI server priamo v prostredí Jupyter Notebook, vytvoriť dva základné endpointy (GET, POST) a otestovať ich pomocou knižnice requests.

- Server sa spúšťa vo vlákne s náhodne prideleným voľným portom, čo zabraňuje konfliktom pri opakovanom spúšťaní
- GET endpoint na / vracia jednoduchú uvítaciu správu
- POST endpoint na /osoba prijíma JSON so štruktúrovanými údajmi o osobe pomocou modelu Pydantic
- Odošlú sa testovacie požiadavky cez requests.get() a requests.post()
- Výstupy sa vypíšu v Jupyter Notebooku

Tento príklad demonštruje, ako kompletne simulovať server aj klienta v rámci jedného skriptu bez potreby externých nástrojov.

**✅ Použité knižnice a nástroje:**
1. FastAPI – webový framework pre tvorbu API
2. uvicorn – spustenie ASGI servera
3. requests – klient na odosielanie HTTP požiadaviek
4. nest_asyncio – podpora asynchrónnych operácií v Jupyteri
5. threading – spustenie servera vo vlastnom vlákne
6. socket – výber voľného portu pre server

**🔗 Dokumentácia:**
1. https://fastapi.tiangolo.com/
2. https://www.uvicorn.org/
3. https://requests.readthedocs.io/
4. https://pydantic-docs.helpmanual.io/
5.  https://docs.python.org/3/library/threading.html
6.  https://docs.python.org/3/library/socket.html

```python
import socket
import nest_asyncio
import threading
import uvicorn
from fastapi import FastAPI
from pydantic import BaseModel
import requests
import time

# 🔧 Funkcia na nájdenie voľného portu
# Definícia funkcie, ktorá nájde voľný (nepoužitý) port na systéme
def get_free_port():
    # Vytvorenie nového socketu (sieťového spojenia)
    s = socket.socket()
    
    # Priradenie socketu na port 0 znamená, že operačný systém vyberie automaticky voľný port
    s.bind(('', 0))
    
    # Získanie čísla portu, ktorý bol automaticky pridelený operačným systémom
    port = s.getsockname()[1]
    
    # Zavretie socketu, aby sa port uvoľnil na ďalšie použitie
    s.close()
    
    # Návrat prideleného portu ako výsledku funkcie
    return port

# ✅ Inicializácia aplikácie a portu
app = FastAPI()
PORT = get_free_port()  # vždy voľný port

@app.get("/")
def root():
    return {"message": "Ahoj z FastAPI v Jupyteri!"}

class Osoba(BaseModel):
    meno: str
    vek: int

@app.post("/osoba")
def vytvor_osobu(osoba: Osoba):
    return {"zprava": f"Osoba {osoba.meno} s vekom {osoba.vek} bola prijatá."}

def spusti_server():
    uvicorn.run(app, host="127.0.0.1", port=PORT)

nest_asyncio.apply()

# ✅ Spustenie servera vo vlákne (vždy nový port = žiadny konflikt)
# Vytvorenie nového vlákna (thread), ktoré bude spúšťať funkciu spusti_server
# Parameter daemon=True zabezpečí, že vlákno sa automaticky ukončí pri ukončení hlavného programu
server_thread = threading.Thread(target=spusti_server, daemon=True)

# Spustenie vlákna – server sa začne spúšťať na pozadí
server_thread.start()

# Pozastavenie hlavného vlákna na 2 sekundy, aby mal server čas sa inicializovať a začať počúvať na porte
time.sleep(2)

# ✅ Test GET
odpoved_get = requests.get(f"http://127.0.0.1:{PORT}/")
print("GET výstup:", odpoved_get.json())

# ✅ Test POST
údaje = {"meno": "Miro", "vek": 135}
odpoved_post = requests.post(f"http://127.0.0.1:{PORT}/osoba", json=údaje)
print("POST výstup:", odpoved_post.json())

# ✅ Info o porte
print(f"FastAPI beží na porte: {PORT}")
```

## 4️⃣a Zistenie dostupnosti portov pre FastAPI server
**📌 Úloha:** Skontrolovať, ktoré porty v rozsahu 8000–8010 sú voľné alebo obsadené na lokálnom počítači.

- Pomocou knižnice socket sa program pokúsi pripojiť na každý port a zistí, či je obsadený alebo voľný
- Funkcia connect_ex() vráti hodnotu 0, ak je port aktívne obsadený nejakým procesom
- Výstupom je textový zoznam so stavom každého portu – vhodné pre testovanie pred spustením uvicorn servera

```python
import socket

def je_port_volny(port):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        return s.connect_ex(('127.0.0.1', port)) != 0

for port in range(8000, 8011):
    status = "voľný" if je_port_volny(port) else "obsadený"
    print(f"Port {port} je {status}")
```

```python
# ✅ Import knižnice socket – poskytuje nízkoúrovňový prístup k sieťovým službám
import socket  # umožňuje vytvárať sieťové spojenia a pracovať s portami (TCP/UDP)

# 🔧 Definícia funkcie na overenie, či je daný port voľný (neobsadený)
def je_port_volny(port):
    # Vytvorenie nového TCP socketu (AF_INET = IPv4, SOCK_STREAM = TCP)
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        
        # Príkaz connect_ex sa pokúsi nadviazať spojenie na daný port
        # Ak vráti 0, port je obsadený. Ak vráti inú hodnotu, port je voľný.
        return s.connect_ex(('127.0.0.1', port)) != 0

# 🔁 Iterácia cez rozsah portov od 8000 po 8010 (vrátane)
for port in range(8000, 8011):
    
    # 🟢 Ak je port voľný, nastaví sa text "voľný", inak "obsadený"
    status = "voľný" if je_port_volny(port) else "obsadený"
    
    # 🖨️ Výpis výsledku pre konkrétny port
    print(f"Port {port} je {status}")
```

## 🔹 HTTP Verbs (metódy)

Používajú sa v HTTP požiadavkách na označenie, akú akciu chcete vykonať s daným zdrojom (napr. v REST API):

- **GET**: získať údaje zo servera  
  *Napr. načítať zoznam používateľov.*

- **POST**: odoslať údaje na server a vytvoriť nový zdroj  
  *Napr. vytvoriť nového používateľa.*

- **PUT**: aktualizovať celý existujúci zdroj na serveri  
  *Napr. prepísať celý používateľský profil.*

- **PATCH**: čiastočne aktualizovať existujúci zdroj  
  *Napr. zmeniť len email používateľa.*

- **DELETE**: odstrániť zdroj zo servera  
  *Napr. vymazať účet používateľa.*

- **TRACE**: vrátiť celú prijatú požiadavku (na debugovanie)  
  *Zriedkavé použitie, najmä v nástrojoch na ladenie siete.*

- **OPTIONS**: ukázať, ktoré HTTP metódy sú podporované pre danú URL  
  *Používa sa napr. pri CORS (Cross-Origin Resource Sharing).*

- **CONNECT**: nadviazať tunel (napr. cez HTTPS)  
  *Používa sa pre zabezpečenú komunikáciu.*

- **PURGE**: zneplatniť cache *(neštandardná metóda)*  
  *Používa sa v CDN systémoch (napr. Varnish).*

- **LOCK**: uzamknúť zdroj pre exkluzívne použitie klientom  
  *Používa sa v protokole WebDAV.*

- **UNLOCK**: odomknúť predtým zamknutý zdroj  
  *Súvisí s LOCK – WebDAV.*

- **MKCOL**: vytvoriť kolekciu (adresár)  
  *Používa sa v systémoch ako WebDAV.*

## 🔹 HTTP Status Codes (stavové kódy)

Číselné kódy odpovedí, ktoré server vracia späť klientovi. Rozdelené do kategórií:

- **1xx – Informational**  
  *Informácie, prechodné stavy (zriedkavé).*

- **2xx – Success**  
  *Požiadavka bola úspešná.*  
  Príklady:  
  - `200 OK`  
  - `201 Created`

- **3xx – Redirection**  
  *Presmerovanie na inú URL.*  
  Príklady:  
  - `301 Moved Permanently`  
  - `302 Found`

- **4xx – Client Errors**  
  *Chyba na strane klienta.*  
  Príklady:  
  - `400 Bad Request`  
  - `404 Not Found`  
  - `422 Unprocessable Entity`

- **5xx – Server Errors**  
  *Chyba na strane servera.*  
  Príklady:  
  - `500 Internal Server Error`  
  - `503 Service Unavailable`

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from typing import Optional

app = FastAPI()

# In-memory "databáza"
users = [
    {"id": 1, "name": "Miro", "email": "miro@vita.sk"},
    {"id": 2, "name": "Jana", "email": "jana@vita.sk"}
]
next_id = 3

# Model pre POST a PUT
class User(BaseModel):
    name: str
    email: str

# Model pre PATCH
class UserUpdate(BaseModel):
    name: Optional[str] = None
    email: Optional[str] = None

# GET: Zobrazenie všetkých používateľov
@app.get("/users", status_code=200)
async def get_users():
    return users

# POST: Vytvorenie nového používateľa
@app.post("/users", status_code=201)
async def create_user(user: User):
    global next_id
    new_user = {"id": next_id, "name": user.name, "email": user.email}
    users.append(new_user)
    next_id += 1
    return {"message": f"User {user.name} created"}

# PUT: Úplná aktualizácia používateľa
@app.put("/users/{user_id}")
async def update_user(user_id: int, user: User):
    for existing_user in users:
        if existing_user["id"] == user_id:
            existing_user["name"] = user.name
            existing_user["email"] = user.email
            return {"message": f"User {user_id} updated to {user.name}"}
    raise HTTPException(status_code=404, detail="User not found")

# PATCH: Čiastočná aktualizácia používateľa
@app.patch("/users/{user_id}")
async def patch_user(user_id: int, user: UserUpdate):
    for existing_user in users:
        if existing_user["id"] == user_id:
            if user.name is not None:
                existing_user["name"] = user.name
            if user.email is not None:
                existing_user["email"] = user.email
            return {"message": f"User {user_id} partially updated"}
    raise HTTPException(status_code=404, detail="User not found")

# DELETE: Vymazanie používateľa
@app.delete("/users/{user_id}", status_code=204)
async def delete_user(user_id: int):
    for i, existing_user in enumerate(users):
        if existing_user["id"] == user_id:
            users.pop(i)
            return None
    raise HTTPException(status_code=404, detail="User not found")
```

## 1️⃣ Autentifikácia používateľa pomocou JWT tokenu
**📌 Úloha:** Implementovať autentifikáciu pomocou OAuth2 Password Flow a vytvoriť chránený endpoint.

- Používateľ získa prístupový token po prihlásení a následne ho použije v hlavičke Authorization.
- Token je šifrovaný pomocou algoritmu HS256 a obsahuje identifikátor používateľa.

**🔗 Dokumentácia:**
- https://fastapi.tiangolo.com/tutorial/security/
- https://fastapi.tiangolo.com/tutorial/security/oauth2-jwt/

```python
# Inštalácia: pip install python-jose passlib[bcrypt]

from fastapi import Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer, OAuth2PasswordRequestForm
from jose import JWTError, jwt
from passlib.context import CryptContext
from datetime import datetime, timedelta

# Tajný kľúč a algoritmus pre JWT
SECRET_KEY = "tajny_kluc_na_signovanie_tokenu"
ALGORITHM = "HS256"
ACCESS_TOKEN_EXPIRE_MINUTES = 30

# Simulovaná databáza používateľov
fake_users_db = {
    "admin": {
        "username": "admin",
        "full_name": "Admin Používateľ",
        "hashed_password": "$2b$12$wXs4k4H99qlzv3Ld/sXjZe88y8i/6r.qQp/04WgPCKOSOTgrQZsI6",  # heslo: tajneheslo
        "disabled": False,
    }
}

# Heslá budeme hašovať pomocou bcrypt
pwd_context = CryptContext(schemes=["bcrypt"], deprecated="auto")

# OAuth2 definícia – očakávame token v hlavičke Authorization: Bearer
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="token")
```

```python
# Pomocné funkcie na prácu s heslami a používateľmi
def verify_password(plain_password, hashed_password):
    return pwd_context.verify(plain_password, hashed_password)

def get_password_hash(password):
    return pwd_context.hash(password)

def get_user(db, username: str):
    if username in db:
        return db[username]
    return None

def authenticate_user(db, username: str, password: str):
    user = get_user(db, username)
    if not user or not verify_password(password, user["hashed_password"]):
        return False
    return user

def create_access_token(data: dict, expires_delta: timedelta | None = None):
    to_encode = data.copy()
    expire = datetime.utcnow() + (expires_delta or timedelta(minutes=15))
    to_encode.update({"exp": expire})
    return jwt.encode(to_encode, SECRET_KEY, algorithm=ALGORITHM)
```

```python
# Endpoint na získanie tokenu (login)
@app.post("/token")
def login(form_data: OAuth2PasswordRequestForm = Depends()):
    user = authenticate_user(fake_users_db, form_data.username, form_data.password)
    if not user:
        raise HTTPException(
            status_code=status.HTTP_401_UNAUTHORIZED,
            detail="Nesprávne prihlasovacie údaje",
            headers={"WWW-Authenticate": "Bearer"},
        )
    access_token = create_access_token(data={"sub": user["username"]})
    return {"access_token": access_token, "token_type": "bearer"}
```

```python
# Získanie aktuálneho používateľa z tokenu
from pydantic import BaseModel

class TokenData(BaseModel):
    username: str | None = None

async def get_current_user(token: str = Depends(oauth2_scheme)):
    credentials_exception = HTTPException(
        status_code=status.HTTP_401_UNAUTHORIZED,
        detail="Neplatný prístupový token",
        headers={"WWW-Authenticate": "Bearer"},
    )
    try:
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        username: str = payload.get("sub")
        if username is None:
            raise credentials_exception
        token_data = TokenData(username=username)
    except JWTError:
        raise credentials_exception
    user = get_user(fake_users_db, username=token_data.username)
    if user is None:
        raise credentials_exception
    return user
```
